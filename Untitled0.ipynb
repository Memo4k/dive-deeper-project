```python
def first_last2(string):
  """Gets the first 2 and last 2 chars of a string.

  Args:
    string: The input string.

  Returns:
    A string containing the first 2 and last 2 chars, or an empty string if the input is too short.
  """

  if len(string) < 2:
    return ""
  else:
    return string[:2] + string[-2:]

# Test cases
print(first_last2('udacity-deci'))  # Output: udci
print (first_last2('deci'))         #output: deci
print (first_last2('w'))            #output: w
print (first_last2('de'))           #output: de
print (first_last2('dec'))          #output: dec


```

```python
import pygame
import time
import random

# إعدادات اللعبة
pygame.init()
width, height = 800, 600
block_size = 20
snake_speed = 15

# ألوان
black = (0, 0, 0)
white = (255, 255, 255)
red = (213, 50, 80)
green = (0, 255, 0)
blue = (50, 153, 213)

# إنشاء النافذة
screen = pygame.display.set_mode((width, height))
pygame.display.set_caption("لعبة الثعبان")

clock = pygame.time.Clock()
font_style = pygame.font.SysFont("bahnschrift", 25)
score_font = pygame.font.SysFont("comicsansms", 35)

def your_score(score):
    value = score_font.render(f"النتيجة: {score}", True, green)
    screen.blit(value, [10, 10])

def our_snake(block_size, snake_list):
    for x in snake_list:
        pygame.draw.rect(screen, black, [x[0], x[1], block_size, block_size])

def message(msg, color):
    mesg = font_style.render(msg, True, color)
    screen.blit(mesg, [width / 6, height / 3])

def gameLoop():
    game_over = False
    game_close = False

    x1, y1 = width / 2, height / 2
    x1_change, y1_change = 0, 0

    snake_List = []
    length_of_snake = 1

    foodx = round(random.randrange(0, width - block_size) / 20.0) * 20.0
    foody = round(random.randrange(0, height - block_size) / 20.0) * 20.0

    while not game_over:
        while game_close:
            screen.fill(blue)
            message("خسرت! اضغط C لإعادة اللعب أو Q للخروج", red)
            your_score(length_of_snake - 1)
            pygame.display.update()

            for event in pygame.event.get():
                if event.type == pygame.KEYDOWN:
                    if event.key == pygame.K_q:
                        game_over = True
                        game_close = False
                    if event.key == pygame.K_c:
                        gameLoop()

        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                game_over = True
            if event.type == pygame.KEYDOWN:
                if event.key == pygame.K_LEFT:
                    x1_change, y1_change = -block_size, 0
                elif event.key == pygame.K_RIGHT:
                    x1_change, y1_change = block_size, 0
                elif event.key == pygame.K_UP:
                    x1_change, y1_change = 0, -block_size
                elif event.key == pygame.K_DOWN:
                    x1_change, y1_change = 0, block_size

        if x1 >= width or x1 < 0 or y1 >= height or y1 < 0:
            game_close = True

        x1 += x1_change
        y1 += y1_change
        screen.fill(blue)
        pygame.draw.rect(screen, green, [foodx, foody, block_size, block_size])

        snake_Head = []
        snake_Head.append(x1)
        snake_Head.append(y1)
        snake_List.append(snake_Head)
        if len(snake_List) > length_of_snake:
            del snake_List[0]

        for x in snake_List[:-1]:
            if x == snake_Head:
                game_close = True

        our_snake(block_size, snake_List)
        your_score(length_of_snake - 1)

        pygame.display.update()

        if x1 == foodx and y1 == foody:
            foodx = round(random.randrange(0, width - block_size) / 20.0) * 20.0
            foody = round(random.randrange(0, height - block_size) / 20.0) * 20.0
            length_of_snake += 1

        clock.tick(snake_speed)

    pygame.quit()
    quit()

gameLoop()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Read the dataset
df = pd.read_csv("hospital_data.csv")

# =====================================
# Step 1: Univariate Analysis
# =====================================

print("Descriptive Statistics:")
print(df[['age', 'stay_days', 'blood_pressure', 'num_prev_visits']].describe())

# Histogram for Stay Days
plt.hist(df['stay_days'], bins=10)
plt.title("Distribution of Stay Days")
plt.xlabel("Stay Days")
plt.ylabel("Frequency")
plt.show()

# Histogram for Age
plt.hist(df['age'], bins=10)
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Boxplot for Stay Days
sns.boxplot(x=df['stay_days'])
plt.title("Boxplot of Stay Days")
plt.show()

# =====================================
# Step 2: Bivariate Analysis
# =====================================

# Scatter Plot: Age vs Stay Days
plt.scatter(df['age'], df['stay_days'])
plt.title("Age vs Stay Days")
plt.xlabel("Age")
plt.ylabel("Stay Days")
plt.show()

# Mean Stay Days by Department
mean_stay = df.groupby('department')['stay_days'].mean()

mean_stay.plot(kind='bar')
plt.title("Average Stay Days by Department")
plt.xlabel("Department")
plt.ylabel("Average Stay Days")
plt.show()

# =====================================
# Step 3: Correlation Analysis
# =====================================

# Convert blood pressure to numeric value
df['systolic_bp'] = df['blood_pressure'].str.split('/').str[0].astype(int)

numeric_cols = ['age', 'stay_days', 'systolic_bp', 'num_prev_visits']

corr_matrix = df[numeric_cols].corr()

print("\nCorrelation Matrix:")
print(corr_matrix)

# Heatmap
plt.figure(figsize=(6,4))
sns.heatmap(corr_matrix, annot=True, cmap='Blues')
plt.title("Correlation Heatmap")
plt.show()

# Find strongest correlation
corr_pairs = corr_matrix.unstack()
corr_pairs = corr_pairs[corr_pairs != 1]

strongest_corr = corr_pairs.abs().idxmax()
corr_value = corr_pairs[strongest_corr]

print("\nStrongest Correlation:")
print(strongest_corr)
print("Correlation Value =", round(corr_value, 3))

# =====================================
# Step 4: Outlier Detection (IQR)
# =====================================

Q1 = df['stay_days'].quantile(0.25)
Q3 = df['stay_days'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['stay_days'] < lower_bound) |
    (df['stay_days'] > upper_bound)
]

print("\nOutliers in Stay Days:")

if len(outliers) == 0:
    print("No outliers found.")
else:
    print(outliers[['patient_id', 'department', 'stay_days', 'age']])

# =====================================
# Step 5: High-Risk Department
# =====================================

cardiology_cases = [
    "Heart Failure",
    "MI",
    "Angina",
    "Arrhythmia",
    "Hypertension"
]

high_risk = df[df['diagnosis'].isin(cardiology_cases)]

print("\nHigh-Risk Patients by Department:")
print(high_risk['department'].value_counts())

print("\nDepartment with Highest Risk Patients:")
print(high_risk['department'].value_counts().idxmax())

# =====================================
# Step 6: Summary
# =====================================

print("\nEDA Summary")

print("1. Most patients stayed between 3 and 8 days, while the maximum stay was 11 days.")

print("2. Older patients tend to have longer hospital stays, showing a positive relationship between age and stay_days.")

print("3. Cardiology appears to admit the highest-risk patients because it includes serious cardiovascular conditions such as Heart Failure, MI, Angina, and Arrhythmia.")
```
